# Mojo factorial

Pick the **Mojo** kernel. Modular notebooks are a Python kernel plus the `%%mojo` cell magic — each Mojo cell is a complete program with `main()`. The booth auto-imports `mojo.notebook`, so you do not need a first-cell `import`.

```python
%%mojo

def main():
    print("Hello from Mojo")
```

```python
%%mojo

def factorial(n: Int) -> Int:
    if n <= 1:
        return 1
    return n * factorial(n - 1)

def main():
    var n = 5
    print(String(n) + "! = " + String(factorial(n)))
```
