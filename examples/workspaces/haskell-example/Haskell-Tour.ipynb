## Values and functions

Definitions persist from cell to cell, like in GHCi.

```haskell
square :: Int -> Int
square x = x * x

map square [1..10]
```

```haskell
factorial :: Integer -> Integer
factorial n = product [1..n]

factorial 25
```

## Laziness

An infinite list is fine as long as you only take part of it.

```haskell
fibs :: [Integer]
fibs = 0 : 1 : zipWith (+) fibs (drop 1 fibs)

take 15 fibs
```

```haskell
primes = sieve [2..] where sieve (p:xs) = p : sieve [x | x <- xs, x `mod` p /= 0]

takeWhile (< 60) primes
```

## Data types and type classes

```haskell
data Shape = Circle Double | Rect Double Double deriving Show

area :: Shape -> Double
area (Circle r) = pi * r * r
area (Rect w h) = w * h

[(s, area s) | s <- [Circle 1, Rect 2 3]]
```

## Libraries

Importing and calling in the same cell used to kill the kernel on GHC 9.10.3; this booth's kernel runs on 9.12.4.

```haskell
import qualified Data.Map as M

wordCounts = M.fromListWith (+) [(w, 1 :: Int) | w <- words "the cat and the hat and the bat"]
M.toList wordCounts
```

```haskell
M.insertWith (+) "cat" 10 wordCounts
```
