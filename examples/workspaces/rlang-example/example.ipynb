# R Example Notebook

A quick demo of R inside CodingBooth.

```r
# Generate sample data
set.seed(42)
x <- 1:20
y <- 2 * x + rnorm(20, sd = 3)

# Fit a linear model
model <- lm(y ~ x)
summary(model)
```

```r
# Plot the data with the regression line
plot(x, y, pch = 19, col = "steelblue",
     main = "Linear Regression",
     xlab = "x", ylab = "y")
abline(model, col = "red", lwd = 2)
```
