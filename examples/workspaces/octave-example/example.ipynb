# Octave Example Notebook

A quick demo of GNU Octave inside CodingBooth.

```octave
% Generate sample data
rand('seed', 42);
x = 1:20;
y = 2 * x + 3 * randn(1, 20);

% Fit a linear model
p = polyfit(x, y, 1);
y_fit = polyval(p, x);

% R-squared
ss_res = sum((y - y_fit).^2);
ss_tot = sum((y - mean(y)).^2);
r_squared = 1 - ss_res / ss_tot;

printf('Slope: %.3f, Intercept: %.3f, R-squared: %.3f\n', p(1), p(2), r_squared);
```

```octave
% Plot the data with the regression line
plot(x, y, 'bo', 'markersize', 8);
hold on;
plot(x, y_fit, 'r-', 'linewidth', 2);
xlabel('x'); ylabel('y');
title('Linear Regression');
legend('Data', 'Fit');
hold off;
```

```octave
% Matrix operations
A = [1 2 3; 4 5 6; 7 8 10];
printf('Matrix A:\n');
disp(A);
printf('Determinant: %.3f\n', det(A));
printf('Eigenvalues:\n');
disp(eig(A));
```
