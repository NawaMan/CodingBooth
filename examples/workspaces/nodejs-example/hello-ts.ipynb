# Hello, TypeScript

The **TypeScript** tslab kernel: every cell is type-checked before it runs. See `type-error.ipynb` for what a mistake looks like.

```typescript
interface Tool {
  name: string;
  stars: number;
}

const tools: Tool[] = [
  { name: "node", stars: 5 },
  { name: "tslab", stars: 4 },
];
```

```typescript
function best(list: Tool[]): Tool {
  return list.reduce((a, b) => (b.stars > a.stars ? b : a));
}

console.log(`best: ${best(tools).name}`);
```

```typescript
const total: number = tools.reduce((n, t) => n + t.stars, 0);
total
```
